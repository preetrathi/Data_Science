## What is Data Analytics?
### Process of inspecting, cleansing, transforming, and modeling data with the goal of discovering the useful information.

## **Steps in Data Analysis Process**
1. Asking Questions
    1. What features will contribute to my analysis?
    2. What features are not important for my Analysis
    3. Which of the features have a strong corelation?
    4. Do I need Data Proprocessing?
    5. What kind of feature manipulation/engineering os required?
2. Data Wrangling/Data Munging/ Data Preprocing
#####     process of transforming and mapping data from one 'raw' data form into another format with the intent of making it more appropriate and valuable.
    1. Gathering Data
      1. CSV
      2. API
      3. Web Scraping
      4. Database
    2. Assessing Data
      1. Finding the number of rows and columns(Shape)
      2. Data types of various columns (info())
      3. Checking for missing values (info())
      4. Check for duplicate data(is_unique)
      5. Memory occupied by dataset(info)
      6. High Level mathematical overview of the data(describe)
    3. Cleaning Data
      1. Missing data
      2. Remove Duplicate Data(drop_duplicates)
      3. Incorrect Data types(astype)

3. Exploratory Data Analysis
    1. Finding Patterns
    2. plotting Graphs
    3. Drived co-relations
    4. Drawing Conclusions
    5.  Communicating Results

      1. Exploring Data
        1. Finding corelation and covariance
        2. Doing univarinat and multivarinat analysis
        3. Plotting graphs(Data Visualization)
      2. Augment(Make changes with rep)
          1. Removing outliers
          2. Merging Dataframes
          3. Adding new column
######        These operations are collectively called as feature engineering

4. Drawing Conclusions
  Machine Learning
  Inferential Statistics
  Descriptive Statistics

5. Communicating Results/Data Storytelling
  1. In Person
  2. Blogspot
  3. Reports
  4. PPTs/Slides decks

## **3. Opening a CSV file from URL**

```python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

## **9. Skip rows/nrows Parameter**

```python
pd.read_csv('data/result.csv', skiprows=[1,2,3], on_bad_lines='skip', engine='python') # for skiping rows
```

```python
# nrows, export n number of rows useful when you are working with large dataset to avoid load millions of rows you can export few of them
pd.read_csv('data/result.csv', nrows=100)
```

## **10. Encoding Parameter**

```python
pd.read_csv('data/result.csv', encoding='latin-1')
```

## **11. Skip Bad Lines**

```python
pd.read_csv('data/result.csv', skip_blank_lines=False) # skip blank lines in csv file

# used to read a csv file with a specific encoding, such as 'utf-8' or 'latin-1', to handle special characters in the data.or parser error
```

## **12. dtypes Parameter**

```python
pd.read_csv('data/result.csv', dtype={'Effective Date': 'Int64'})
```

## **13. Handling Dates**

```python
pd.read_csv('data/result.csv', parse_dates=['Date of Birth', 'Place of Birth']).info()
```

## **14. Convertors**

```python
pd.read_csv('data/result.csv', encoding='utf-8')
```

```python
def rename_col(name):
    if name == 'National':
        return 'Local'
    else:
        return name
```

```python
pd.read_csv('data/result.csv', encoding='utf-8', converters={'Entity Level': rename_col})
```

## **15. na_values Parameter**

```python
# Use to specifiy which kind of values treated as missing values, by default it is set to NaN but you can change it to any value.

# pd.read_csv('data/result.csv', na_values=['NA', 'Missing', 'Not Available'])

pd.read_csv('data/result.csv', na_values={'Entity Level': ['NA', 'Missing'], 'Place of Birth': ['Not Available']})
```

## **16. Loading a huge dataset in chunks**

```python
df_chunks = pd.read_csv('data/result.csv', chunksize=50000)
```

```python
for chunk in df_chunks:
    print(chunk.shape)
```
